# 05 — Model Training

**Purpose:** Train LR, SVM, RF with subject-level cross-validation.

```python
import sys, os
PROJECT_ROOT = os.path.dirname(os.getcwd()) if os.path.basename(os.getcwd()) == "notebooks" else os.getcwd()
sys.path.insert(0, PROJECT_ROOT)
os.chdir(PROJECT_ROOT)
import pandas as pd
from src.dataset import prepare_train_test_data
subject_df = pd.read_csv('data/features/subject_features.csv')
X_train, X_test, y_train, y_test, feat_names, groups = prepare_train_test_data(subject_df, test_size=0.2, random_state=42)
print(f'Train: {len(X_train)}, Test: {len(X_test)}, Features: {len(feat_names)}')
```

```python
from src.train import train_and_compare
configs = [
    {'name':'logistic_regression','params':{'max_iter':1000,'random_state':42,'class_weight':'balanced'}},
    {'name':'svm','params':{'kernel':'rbf','random_state':42,'class_weight':'balanced','probability':True}},
    {'name':'random_forest','params':{'n_estimators':200,'random_state':42,'class_weight':'balanced'}}
]
results = train_and_compare(X_train, y_train, X_test, y_test, configs, groups=groups, cv_folds=5)
for r in results:
    print(f'{r.name}: f1={r.test_metrics["f1_macro"]:.3f}')
```

```python
from src.train import select_best_model, save_model
best = select_best_model(results, 'f1_macro')
print(f'Best: {best.name}')
save_model(best, 'models', feat_names, {'A':'Alzheimers','F':'FTD','C':'Control'}, {})
```
