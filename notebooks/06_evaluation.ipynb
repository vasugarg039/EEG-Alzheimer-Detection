# 06 — Model Evaluation

**Purpose:** Compute metrics and visualize results.

```python
import sys, os
PROJECT_ROOT = os.path.dirname(os.getcwd()) if os.path.basename(os.getcwd()) == "notebooks" else os.getcwd()
sys.path.insert(0, PROJECT_ROOT)
os.chdir(PROJECT_ROOT)
import pandas as pd, numpy as np
from src.train import load_model
model, metadata = load_model('models/random_forest_20261017_024400')
subject_df = pd.read_csv('data/features/subject_features.csv')
X = subject_df[metadata['feature_names']].values
y = subject_df['group'].values
X = np.nan_to_num(X)
y_pred = model.predict(X)
print(f'Predictions: {len(y_pred)}')
```

```python
from src.evaluate import compute_metrics, generate_classification_report
metrics = compute_metrics(y, y_pred, labels=['A','F','C'])
print(f'Accuracy: {metrics["accuracy"]:.4f}')
print(f'Bal Acc: {metrics["balanced_accuracy"]:.4f}')
print(f'F1 Macro: {metrics["f1_macro"]:.4f}')
print(generate_classification_report(y, y_pred, labels=['A','F','C']))
```

```python
import matplotlib.pyplot as plt
from src.visualization import plot_confusion_matrix
plot_confusion_matrix(y, y_pred, labels=['A','F','C'])
plt.show()
```
