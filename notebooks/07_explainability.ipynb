# 07 — Model Explainability

**Purpose:** Understand which features drive predictions.

> Feature importance reflects model behavior, not biological causation.

```python
import sys, os
PROJECT_ROOT = os.path.dirname(os.getcwd()) if os.path.basename(os.getcwd()) == "notebooks" else os.getcwd()
sys.path.insert(0, PROJECT_ROOT)
os.chdir(PROJECT_ROOT)
import pandas as pd, numpy as np
from src.train import load_model
model, metadata = load_model('models/random_forest_20261017_024400')
subject_df = pd.read_csv('data/features/subject_features.csv')
X = subject_df[metadata['feature_names']].values
y = subject_df['group'].values
X = np.nan_to_num(X)
```

```python
from src.explainability import compute_permutation_importance
imp = compute_permutation_importance(model, X, y, metadata['feature_names'], n_repeats=5)
print('Top 10 features:')
for i in range(10):
    print(f'  {imp["ranked_features"][i]}: {imp["ranked_importance"][i]:.4f}')
```

```python
import matplotlib.pyplot as plt
from src.visualization import plot_feature_importance
plot_feature_importance(imp, top_n=15)
plt.show()
```
