# 03 — EEG Segmentation (Epoching)

**Purpose:** Split continuous EEG into fixed-length epochs with overlap.

```python
import sys, os
PROJECT_ROOT = os.path.dirname(os.getcwd()) if os.path.basename(os.getcwd()) == "notebooks" else os.getcwd()
sys.path.insert(0, PROJECT_ROOT)
os.chdir(PROJECT_ROOT)
from src.config import load_config
from src.data_loader import EEGDataset
from src.preprocessing import EEGPreprocessor
from src.segmentation import EEGSegmenter
config = load_config()
dataset = EEGDataset(config)
dataset.load_metadata()
preprocessor = EEGPreprocessor(config['preprocessing'])
segmenter = EEGSegmenter(config['segmentation'])
print('Segmenter ready')
```

```python
raw = dataset.load_eeg('sub-001')
preproc = preprocessor.preprocess(raw)
seg = segmenter.segment(preproc.raw_filtered)
print(f'Total: {seg.n_epochs_total}, Clean: {seg.n_epochs_clean}, Rejected: {seg.n_epochs_rejected}')
```

```python
import matplotlib.pyplot as plt
data = seg.epochs.get_data()
fig, ax = plt.subplots(3,1,figsize=(10,6))
for i in range(min(3,data.shape[0])):
    ax[i].plot(data[i,0])
plt.tight_layout(); plt.show()
```
